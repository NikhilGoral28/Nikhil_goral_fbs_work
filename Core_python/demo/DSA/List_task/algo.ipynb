```python
#linearSearch
#timecomplexity - O(n) - worst case
# O(1) - best case

def linearSearch(Ele):
    for i in range(0,len(l)):
        if l[i] == Ele:
            return i
    else:
        return -1
        
l = [20,30,40,50,60,70,80]

ele = int(input("Enter element: "))

res = linearSearch(ele)

if res != -1:
    print(f"Element {ele} present at index {res}")
else: 
    print(f"element {ele} not present in the list")
```
