```python
#linearSearch
#timecomplexity - O(n) - worst case
# O(1) - best case

def linearSearch(Ele):
    for i in range(0,len(l)):
        if l[i] == Ele:
            return i
    else:
        return -1
        
l = [20,30,40,50,60,70,80]

ele = int(input("Enter element: "))

res = linearSearch(ele)

if res != -1:
    print(f"Element {ele} present at index {res}")
else: 
    print(f"element {ele} not present in the list")
```

```python
def bisort(l,ele):

    start = 0
    end = len(l) -1

    while start <= end:

        mid = (start+end) // 2

        if l[mid] == ele:
            return mid
        elif ele < l[mid]:
            end = mid-1
        elif ele > l[mid]:
            start = mid +1
    else:
        return -1
    

l = [2,3,4,5,6,7,8,9]

ele = 7

print(bisort(l,ele))
```

```python
#bubble sort
#time complexity - O(n) - best case - where the list is already sorted
#time complexity - o(n^2) worst case
def bubblesort(l):
    size = len(l)

    for i in range(1,size):  #outer for loop for the elements
        for j in range(0,size-i): #inner for loop for the camparison
            if l[j] > l[j+1]:

                l[j],l[j+1] = l[j+1],l[j]


l = [60,40,34,56,75,32]

bubblesort(l)
print(l)

```

```python
#selection sort

def selectionsort(li):

    size = len(li)

    for i in range(0,size-1):
        min = i
        for j in range(i+1, size):
            if (li[min] > li[j]):
                min = j
        li[i], li[min] = li[min], li[i]
        print(li)

li = [23,43,54,65,76,7,87]

selectionsort(li)
```
