```python
#mean
li = [40,30,60,80,70]
n = len(li) 
t = 0
for i in range(0,len(li)):
    t += li[0]

mean = t/n
print(mean)
```

```python
#min

l = [20,30,40,50,60,70]

mn = l[0]
smax = 0

for i in range(1,len(l)):
    if l[i] > mn:
        smax = mn
        mn = l[i]
    elif l[i] > smax:
        smax = l[i]

print(mn)
print(smax)
```

```python
#slicing

li = [1,2,3,4,5,6,7,8,9]

#li[start:stop:step]
#stop: end -1
#step by default 1

print(li[0:4])

print(li[0:])
print(li[0:7:2])
```
