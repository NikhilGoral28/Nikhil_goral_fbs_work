```python
#fibonacci series upto n

def fibo(n,a,b):
    for i in range(1,n+1):
        c = a+b
        print(c,end=' ')
        a=b
        b=c

num = int(input("How many fibonacci number you want: "))
fibo(num,-1,1)
```

```python
#type 2
#wap to print all prime number upto n

def prime(n):
    for num in range(2, n + 1):
        is_prime = True
        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=" ")

n = int(input("Enter number: "))
prime(n)
```
