```python
#recuresive function

def sumOfSeries(n):
    if (n==0):
        return 0
    else:
        return n + sumOfSeries(n-1)


n = 1002
res = sumOfSeries(n)

print(res)
```

```python
#factorial of number using recursion

def fact(n):
    if n ==0:
        return 1
    return n*fact(n-1)

print(fact(5))
```

```python
#wap to seperate digits from given number

def seperate(n):
    if n < 10:
        print(n)
        return
    print(n%10)
    seperate(n//10)


print(seperate(12))
```

```python
#sum of digit

def sepDigit(num):
    if num == 0:
        return 0
    else:
        return num % 10 + sepDigit(num//10)
    

num = 334343
res=(sepDigit(num))
print(res)
```

```python
#reverse a number using recursive function

def reverse(num, rev=0):
    if num == 0:
        return rev
    return reverse(num//10, rev*10 + num%10)

print(reverse(3432))
    
```

```python
def reverse_num(num):

    rev = 0

    while num > 0:
        digit = num %10
        num = num // 10

        rev = digit + rev *10

    return rev

print(reverse_num(7654))
```
