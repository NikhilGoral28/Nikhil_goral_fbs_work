```python
#deafult values

def add(a, b=10):
    c = a +b

    return c


a = 20

print(add(a))

#can be replace 

a = 30
b = 40

print(add(a,b))
```

```python
#keyword default 

def employee(id, name, city, salary = 15000):
    print("id: ", id)
    print("name: ", name)
    print("city: ", city)
    print("salary: ", salary)


employee(name = "Adi", city="Kolhapur", salary = 15000 * 1.05, id=1)
```

```python
def add (*num):

    return sum(num)

print(add(4,5,6,7))
```

```python
#**kargs

def employee(**data):
    for key in data.items():
        print(key)

employee(name = "Adi", city="Kolhapur", salary = 15000 * 1.05, id=1)
```
