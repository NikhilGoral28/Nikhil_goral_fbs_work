```python
num = int(input("Enter a number: "))

sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num = num //10

print("Sum of given number digit: ", sum)
```

```python
num = int(input("Enter Number: "))
sum = 0

for i in range(1,num):
    if num % i == 0:
        sum += i 

if sum == num:
    print("Number is perfect number")
else:
    print("Number is not perfect")
```
