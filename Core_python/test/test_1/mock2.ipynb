```python
##write a program to reverse three-digit number. 


n = 456

rev = 0

while n > 0:
    digit = n % 10

    n = n // 10

    rev = rev*10 + digit


print(rev)
```

```python
##Write a program to check if given 3 digit number is a palindrome or not

num = 121


if num == int(str(num)[::-1]):
    print("palindrome")
else:
    print("not")
```

```python
## WAP to check if a given number is prime number or not.

n = 56

for i in range(2,n):
    if n % i == 0:
        print("not a prime number")
        break
    else:
        print("number is prime")
```

```python
##WAP to print all integers upto n that aren’t divisible by 2 and 3.

n = 30

for i in range(1,n+1):
    if i % 2 == 0 or i % 3 == 0:
        continue
    else:
        print(i)
```

```python
#armstrong

num = 156
n = len(str(num))
temp = num
total = 0

while num > 0:
    digit = num %10
    num = num //10

    total += digit ** n

if total == temp:
    print("armstrong")

```

```python
##WAP to print Armstrong number within a given range

lw = 150
up = 160

for num in range(lw,up+1):

    n = len(str(num))

    total = 0
    temp = num
    while temp > 0:
        digit = temp % 10

        total += digit ** n

        temp = temp // 10
    

    if total == num:
        print(num)

```

```python
num = int(input("Enter a number: "))

temp = num 
sum = 0

# count number of digits
n = len(str(num))
while num > 0:
    digit = num % 10
    sum = sum + digit ** n
    num = num // 10
    
if sum == temp:
    print(f"{temp} is an Armstrong number.")
else:
    print("number is not armstrong number")
```
