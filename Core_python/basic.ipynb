```python
sen = input("Enter a sentence: ")

words = sen.split()
rev = []

for word in words:
    rev.append(''.join(reversed(word)))  # convert iterator to string

print(rev)

# join the reversed words back into a sentence
rev_sentence = ' '.join(rev)
print(rev_sentence)

    
```
